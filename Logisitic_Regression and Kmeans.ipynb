KMeans
    -Unsupervised Algorithm
    -Unlabelled data
    -Consists of Only Input
    -Clustering algorithm
    -No target variable(O/P)
    -Group similar data points together

    K-No of Clusters(How many group we can make based on data points)
    Means-Calculate  the average point to find centroid(Point)

    Goal:
        Within the data points in cluster is similar /Close to each other
        B/w different cluster data point should far away

        Ex:
        Customer Mall Segmenation

        Income            Spending
        10                 10 
        20                 15
        30                 20
        22                 4
        35                 23
        40                 10

        Centroid -Randomly I choose One Centroid point
        (10,10)-Centroid Point For cluster 1
        (20,15)-Centroid Point For CLuster 2

        Distance Metrics-Eucledian Distance
        Sqrt(x2-x1)**2+(y2-y1)**2

        x1,y1-(10,10)
        x2,y2-Centroid Point(10,10)

        D.M=0->Cluster 1

        x1,y1=(20,15)-Centroid 
        x2,y2=(10,10)
        =sqrt(20-10)**2+(15-10)**2
        =100+25
        125
        =11.4 ->Cluster 2

        CLuster 1         CLuster 2
        10,10              20,15
        30,8               15,19

        Within this data point find average of x,y we will get new centroid
        again compare all the input with centroid points if there is change we need to update the centroid and
        find the distance which is closer to that point until that we need to do this process repeatedly

        Otherwise if ur data is not changing we got the perfect cluster

        Cluster 1-High Income Vs High Spending
        Cluster 2-Low Income Vs Low Spending

        In ur model we want to know ur data points how many clusters we can make to know that we have the concepts
        Elbow Method
        WCSS(Within CLuster Sum oF Square)

        Ur data where it will form the elbow that point u can number of cluster

        Quality of cluster is good or not we have in this silhoutte score
        

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "Income":[20,22,25,30,32,35,40,45,50,55],
    "Spending":[10,3,4,20,10,18,2,4,12,10]})
```

```python
df
```

```python
plt.scatter(df["Income"],df["Spending"],color="green")
plt.grid()
plt.show()
```

```python
X=df[["Income","Spending"]]
```

```python
X
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,5):
    km=KMeans(n_clusters=i)
    km.fit(X)#Unlabelled data contains only input
    wcss.append(km.inertia_) #Inertia point we get to in which cluster we can decide to form groups
plt.plot(range(1,5),wcss,marker='o') #Data point circle
plt.grid(True)
plt.show()
```

```python
km=KMeans(n_clusters=3)
km.fit(X)
cluster=km.predict(X)
```

```python
df['Cluster']=cluster
```

```python
plt.scatter(df["Income"],df["Spending"],c=df["Cluster"],cmap="viridis")
plt.grid()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(X,cluster)
```

```python
x=df.drop("Outcome",axis=1)
y=df['Outcome']
```

```python
#Train test and split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
#Train data is 75% and test data is 25%
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(x_train,y_train)
```

```python
pred_linear=reg.predict(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
log=LogisticRegression()
```

```python
#Assign the target variable
log.fit(x_train,y_train)
```

```python
pred_outcome=log.predict(x_test) #Probability output using sigmoid function
#And final outcome decide by threshold value
```

```python
pred_outcome
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,pred_outcome)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
accuracy_score(y_test,pred_outcome)
```
